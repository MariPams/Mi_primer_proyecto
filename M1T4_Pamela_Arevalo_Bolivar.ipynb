## Ejercicio 2. Operaciones con strings

El operador '+' concatena cadenas de texto, es decir las une. El operador '*' repite una cadena el numero de veces indicado

```python
Nombre = "Emmanuel"
Apellido = "Delgado"

print(Nombre + " " + Apellido)
print(Nombre + "" + Apellido + ".")
print(Nombre * 3)
```

## Ejercicios de operaciones y expresiones 

1. Media aritmetica: Calcula la suma de todas las variables numericas y divide el resultado entre el numero de variables. Los parentesis permiten realizar la suma antes de la division. 

```python
Edad_padre = 65
Edad_hermano = 40
Edad_hermana = 41
Edad_sobrino = 12

Media_edad = (Edad_padre + Edad_hermano + Edad_hermana + Edad_sobrino) / 4
print(" La edad media familiar es: " + str(Media_edad))
```

2. Redondeo: La funcion round permite redondear un numero a la cantidad de decimales indicadas.

```python
resultado = (365/12) * 14.7
resultado_redondeado = round(resultado, 2)

print("El resultado original es: " + str(float(resultado)))
print("El resultado redondeado es: " + str(resultado_redondeado))
```

3. Operadores lógicos: Permiten combinar o negar condiciones y obtener resultados en booleano

```python
Username = "mariaareva"
Password = "Tokio"

Caracteres_usuario = len(Username)
Condicion_usuario = 3 <= len(Username) <= 10
Condicion_password = Password == "Tokio" or Password == "Python"

Datos_validos = Condicion_usuario and Condicion_password

print("Numero de caracteres de usuario: ", Caracteres_usuario)
print("Usuario valido: ", Condicion_usuario)
print("Password valido: ", Condicion_password)
print("Ambas condiciones se cumplen: ", Datos_validos)
```

## Ejercicio de Input/Output

Los operadores de asignacion permiten realizar una operacion y actualizar el valor de una variable

```python
Numero = 15
Cantidad = -3 

Numero += Cantidad 
print("Numero resultante: ", Numero)

Numero -= Cantidad
print("Numero resultante: ", Numero)

Numero *= Cantidad
print("Numero resultante: ", Numero)

Numero /= Cantidad 
print("Numero resultante: ", Numero)
```

** Cada operacion usa el resultado del paso anterior

2. Funcion Input(): Devuelve una cadena de texto, aunque se introduzcan numeros. 

```python
try:
    nombre = str(input("Ingrese tu nombre: "))
    print("Hola, ", nombre + "!", type(nombre))
    
    edad = int(input("Ingrese tu edad: "))
    print("Tu edad es: ", edad, type(edad))
    
    altura = float(input("Ingrese tu altura en metros: "))
    print("Tu altura es: ", altura, type(altura))
    
except ValueError:
    print("Error: Debes ingresar un número válido para la edad y la altura.")
    
```

3. Formateo de numeros: Permite controlar como se muestran los numeros. Cambia su presentacion pero no el valor de la variable

```python
Numero = 12 
Numero_decimal = 3.1416

print("Entero con 5 digitos rellenados con ceros: {:05d}".format(Numero))
print("Número decimal con 5 dígitos en la parte entera y 3 en la parte decimal:  {:09.3f}".format(Numero_decimal))
```

4. Formas de formateo: 

```python
try:
    Nombre = str(input("Ingresa tu nombre: "))
    Altura = float(input("Ingresa tu altura en metros: "))
    Peso = float(input("Ingresa tu peso en kilogramos: "))
    
    Mensaje_indices = "Hola, {0}! Tu altura es {1:.2f} metros y tu peso es {2:.3f} kilogramos.".format(Nombre, Altura, Peso)
    Mensaje_variables = "Hola, {Nombre}! Tu altura es {Altura:.2f} metros y tu peso es {Peso:.3f} kilogramos.".format(Nombre=Nombre, Altura=Altura, Peso=Peso)
    Mensaje_posicion = "Hola, tu peso es {0:.3f} kilogramos, tu altura es {1:.2f} metros, {2}. Te veo bien!".format(Peso, Altura, Nombre)    
    
    print(Mensaje_indices)
    print(Mensaje_variables)
    print(Mensaje_posicion)
    
except ValueError:
    print("Error: Debes ingresar un número válido para la altura y el peso.")   
```
